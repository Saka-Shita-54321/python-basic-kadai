```python
class Human:
    #コンストラクタを定義
    def __init__(self):
        self.name = ""
        self.age = ""

    #メソッドを定義
    def set_name(self, name, age):
        self.name = name
        self.age = age

    def printinfo(self):
        print(self.name,self.age)

# インスタンス化
profile = Human()

#メソッドにアクセスして実行する
profile.set_name("やまだ","33")
profile.printinfo()
```
