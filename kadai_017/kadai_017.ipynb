```python
class Human:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def check_adult(self):
        if self.age >= 20:
            print(f"{self.name}は大人です")
        else:
            print(f"{self.name}は大人ではありません")

humans = [Human("山田",30), Human("松村",31), Human("たら",3)]

count = len(humans)

for human in humans:
    human.check_adult()
```
